```python
from typing import TypedDict
from langgraph.graph import START, END, StateGraph
from langchain_groq import ChatGroq

# Define the state dict structure.
class QAState(TypedDict):
    question: str
    ans: str

# Define the node function.
def llmQA(state: QAState) -> QAState:
    llm = ChatGroq(model="llama-3.1-8b-instant")
    response = llm.invoke(state["question"])
    state["ans"] = response.content
    return state

# Define the graph.
graph = StateGraph(QAState)

# Add nodes & edges to the graph.
graph.add_node("llmQA", llmQA)

graph.add_edge(START, "llmQA")
graph.add_edge("llmQA", END)

# Compile.
workflow = graph.compile()

# Invoke.
initial_state = {"question": "Explain langgraph"}
final_state = workflow.invoke(initial_state)
print(final_state)
```
