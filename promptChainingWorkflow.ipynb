```python
from typing import TypedDict
from langgraph.graph import START, END, StateGraph
from langchain_groq import ChatGroq

llm = ChatGroq(model="llama-3.1-8b-instant")
# Define the state dict structure.
class BlogState(TypedDict):
    topic: str
    outline: str
    blog: str

# Define the node function.
def generateOutline(state: BlogState) -> BlogState:
    
    question=f"Generate a outline for the topic {state['topic']}"
    outline = llm.invoke(question)
    state["outline"] = outline.content
    return state

def generateBlog(state: BlogState) -> BlogState:
    
    question=f"Generate a blog for the outline {state['outline']}"
    response = llm.invoke(question)
    state["blog"] = response.content

    return state

# Define the graph.
graph = StateGraph(BlogState)

# Add nodes & edges to the graph.
graph.add_node("generateOutline", generateOutline)
graph.add_node("generateBlog", generateBlog)

graph.add_edge(START, "generateOutline")
graph.add_edge("generateOutline", 'generateBlog')
graph.add_edge("generateBlog", END)

# Compile.
workflow = graph.compile()

# Invoke.
initial_state = {"topic": "langgraph"}
final_state:BlogState = workflow.invoke(initial_state)
print(final_state["blog"])
```
